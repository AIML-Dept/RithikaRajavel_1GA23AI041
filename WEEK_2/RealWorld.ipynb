# Week 2 - Real World: Quantum Phase Error

### Objective

Simulate a simple quantum communication error by applying a
random phase error to a qubit.

We will first create a superposition state and then apply a
random phase rotation. The resulting state will be visualized
on the Bloch sphere.

```python
import random
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Step 1: Create a Superposition State

The qubit starts in the |0⟩ state.

We apply a Hadamard gate to create an equal superposition:

|+⟩ = (|0⟩ + |1⟩) / √2

```python
qc = QuantumCircuit(1)

qc.h(0)

print(qc)
```

## Step 2: Generate a Random Phase Error

In a real quantum communication system, noise can change
the phase of a qubit.

Here we generate a random phase between 0 and 2π to simulate
this type of error.

```python
phase_error = random.uniform(0, 2 * 3.14159)

print(f"Random Phase Error: {phase_error:.3f} radians")
```

## Step 3: Apply the Phase Error

The RZ gate rotates the phase of the qubit around the Z-axis
of the Bloch sphere.

The randomly generated phase is used as the rotation angle.

```python
qc.rz(phase_error, 0)

print(qc)
```

## Step 4: Obtain the Final Statevector

We calculate the statevector after the phase error has been applied.

```python
state = Statevector.from_instruction(qc)

print("Final Statevector:")
print(state)
```

```python
state.draw("bloch")
```

## Conclusion

A random phase error was introduced to a qubit in a
superposition state.

The RZ gate simulated the phase error by rotating the
qubit around the Z-axis of the Bloch sphere.

This demonstrates how phase noise can affect quantum
information during communication.
