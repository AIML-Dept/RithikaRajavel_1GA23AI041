# Week 2 - Easy: Single Qubit Statevector

### Objective
Create a qubit in the |0⟩ state, apply an X gate, and visualize
the resulting statevector.

The X gate changes the state |0⟩ to |1⟩.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Step 1: Create the Qubit

A single qubit is initially in the computational basis state |0⟩.

```python
qc = QuantumCircuit(1)

print(qc)
```

## Step 2: Apply the X Gate

The X gate acts like a quantum NOT gate.

It changes:

|0⟩ → |1⟩

```python
qc.x(0)

print(qc)
```

## Step 3: Obtain the Statevector

The statevector represents the quantum state of the qubit
after applying the X gate.

```python
state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)
```

```python
state.draw("qsphere")
```

## Conclusion

The qubit was initially in the |0⟩ state. After applying the X gate,
the qubit changed to the |1⟩ state.

Therefore:

|0⟩ → X → |1⟩
