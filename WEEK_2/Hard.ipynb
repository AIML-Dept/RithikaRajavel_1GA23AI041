# Week 2 - Hard: Two-Qubit Superposition

### Objective

Create a 2-qubit quantum circuit in an equal superposition
and verify the amplitudes and probabilities of all possible states.

For 2 qubits, there are 4 possible states:

|00⟩, |01⟩, |10⟩, |11⟩

Each state should have an equal probability of approximately 1/4.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
qc = QuantumCircuit(2)

print(qc)
```

## Step 2: Apply Hadamard Gates

A Hadamard gate creates an equal superposition.

Applying H to both qubits produces an equal superposition
of all four possible two-qubit states.

```python
qc.h(0)
qc.h(1)

print(qc)
```

## Step 3: Calculate the Statevector

The statevector shows the amplitude associated with each
possible two-qubit state.

```python
state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)
```

## Step 4: Calculate the Probabilities

The probability of a state is the square of the magnitude
of its amplitude.

Since each amplitude is 0.5:

0.5² = 0.25

Therefore, each state should have a probability of 25%.

```python
probabilities = state.probabilities_dict()

print("State Probabilities:")

for state_name, probability in probabilities.items():
    print(f"{state_name}: {probability:.3f}")
```

## Step 5: Verify the Total Probability

The probabilities of all possible states must add up to 1.

This confirms that the quantum state is properly normalized.

```python
total_probability = sum(probabilities.values())

print(f"Total Probability: {total_probability:.3f}")
```

## Conclusion

Applying a Hadamard gate to both qubits creates an equal
superposition of the four possible states.

Each state has an amplitude of 0.5 and a probability of 0.25.

Therefore:

|00⟩ = 25%
|01⟩ = 25%
|10⟩ = 25%
|11⟩ = 25%

The total probability is 1, confirming that the state is normalized.
