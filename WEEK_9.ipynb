# Tutorial 9 – Easy
## Φ⁺ Bell State

The Bell state Φ⁺ is

|Φ⁺⟩ = (|00⟩ + |11⟩) / √2

When measured, only 00 and 11 should occur.

```python
import sys
!{sys.executable} -m pip install qiskit
!{sys.executable} -m pip install qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure([0, 1], [0, 1])

result = AerSimulator().run(qc, shots=1024).result()
print(result.get_counts())
```

# Tutorial 9 – Medium
## Four Bell States

The four Bell states are:

Φ⁺ = (|00⟩ + |11⟩) / √2
Φ⁻ = (|00⟩ - |11⟩) / √2
Ψ⁺ = (|01⟩ + |10⟩) / √2
Ψ⁻ = (|01⟩ - |10⟩) / √2

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

states = {
    "Phi+": [],
    "Phi-": [],
    "Psi+": [],
    "Psi-": []
}

for name in states:
    qc = QuantumCircuit(2, 2)

    if "Psi" in name:
        qc.x(1)

    qc.h(0)
    qc.cx(0, 1)

    if name in ["Phi-", "Psi-"]:
        qc.z(0)

    qc.measure([0, 1], [0, 1])

    counts = sim.run(qc, shots=1024).result().get_counts()
    states[name] = counts

for name, counts in states.items():
    print(name, ":", counts)
```

# Tutorial 9 – Hard
## Bell State in the Hadamard Basis

Instead of measuring directly in the computational basis,
we apply a Hadamard gate before measurement.

This changes the measurement basis and reveals different
correlation patterns.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Create Phi+
qc.h(0)
qc.cx(0, 1)

# Change to Hadamard basis
qc.h(0)
qc.h(1)

qc.measure([0, 1], [0, 1])

result = AerSimulator().run(qc, shots=1024).result()
print(result.get_counts())
```

# Tutorial 9 – Real World
## Entanglement and Quantum Key Distribution

Entanglement can be used as a resource in quantum communication.

In an entanglement-based QKD scheme:

1. Two parties share entangled qubits.
2. They measure their qubits using selected bases.
3. Correlated results can be used to establish shared information.
4. Comparing a subset of results can reveal disturbances caused by an eavesdropper.

The strong correlations of entangled states therefore provide
a way to detect interference with the communication process.

## GHZ State and W State

A 3-qubit GHZ state is

(|000⟩ + |111⟩) / √2

A 3-qubit W state is

(|001⟩ + |010⟩ + |100⟩) / √3

We compare their measurement distributions.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

# GHZ state
ghz = QuantumCircuit(3, 3)
ghz.h(0)
ghz.cx(0, 1)
ghz.cx(0, 2)
ghz.measure([0, 1, 2], [0, 1, 2])

# W state
w = QuantumCircuit(3, 3)

# Prepare W state
w.ry(2 * __import__("math").acos(1 / __import__("math").sqrt(3)), 0)
w.cx(0, 1)
w.ry(2 * __import__("math").acos(1 / __import__("math").sqrt(2)), 1)
w.cx(1, 2)
w.cx(0, 1)

w.measure([0, 1, 2], [0, 1, 2])

print("GHZ:", sim.run(ghz, shots=1024).result().get_counts())
print("W:", sim.run(w, shots=1024).result().get_counts())
```
