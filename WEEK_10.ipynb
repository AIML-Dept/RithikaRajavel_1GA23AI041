# Tutorial 10 – Easy
## Superdense Coding

Superdense coding uses a shared entangled pair to send
two classical bits by transmitting only one qubit.

```python
!pip install qiskit qiskit-aer -q

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Create shared Bell pair
qc.h(0)
qc.cx(0, 1)

# Encode message 00 (do nothing)

# Receiver decodes
qc.cx(0, 1)
qc.h(0)

qc.measure([0, 1], [0, 1])

result = AerSimulator().run(qc, shots=1024).result()
print(result.get_counts())
```

# Tutorial 10 – Medium
## Encode and Decode All Messages

Test the messages 00, 01, 10 and 11.
The decoded output should match the original message.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

for message in ["00", "01", "10", "11"]:
    qc = QuantumCircuit(2, 2)

    # Create entanglement
    qc.h(0)
    qc.cx(0, 1)

    # Encode message
    if message[1] == "1":
        qc.x(0)
    if message[0] == "1":
        qc.z(0)

    # Decode
    qc.cx(0, 1)
    qc.h(0)
    qc.measure([0, 1], [0, 1])

    counts = sim.run(qc, shots=1024).result().get_counts()
    print(message, "->", counts)
```

# Tutorial 10 – Challenge
## Superdense Coding with Noise

Apply a bit-flip error randomly with probability p.
Estimate the decoding error rate using repeated simulation.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, pauli_error

p = 0.1

noise = NoiseModel()
noise.add_all_qubit_quantum_error(
    pauli_error([("X", p), ("I", 1-p)]), ["id"]
)

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)

# Encode 00
qc.id(0)  # Noisy channel

qc.cx(0, 1)
qc.h(0)
qc.measure([0, 1], [0, 1])

result = AerSimulator(noise_model=noise).run(
    qc, shots=1000
).result()

counts = result.get_counts()
errors = 1000 - counts.get("00", 0)

print("Counts:", counts)
print("Estimated decoding error rate:", errors / 1000)
```
